```python
import sys
from pathlib import Path
import os

project_root=Path("..").resolve()

os.chdir(project_root)
sys.path.append(str(project_root))

from rag_modular.chunking.hierarchy_breadcrumb import load_document, hierarchy_breadcrumb_chunking
```

```python
documents=load_document()

print(f'source: {documents[0]["source"]}\ncontent: {documents[0]["content"]}')
```

```python
chunks=[]
chunk_id=0

for doc in documents:
    doc_chunks=hierarchy_breadcrumb_chunking(doc["content"], chunk_size=512)
    for c in doc_chunks:
        chunks.append({
        "chunk_id":chunk_id,
        "breadcrumb":c["breadcrumb"],
        "content":c["content"]
        })
    chunk_id+=1

for i in chunks[0:3]:
    print(f'''
        chunk_id: {i["chunk_id"]}
        breadcrumb: {i["breadcrumb"]}
        content :{i["content"]}''')
```
